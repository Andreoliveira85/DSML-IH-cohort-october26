## 1. Introduction to OOP (10 min)

**What is OOP?**  
Object-Oriented Programming is a paradigm that models software as a collection of “objects,” each encapsulating data (attributes) and operations on that data (methods).  

**Key Pillars of OOP:**  
- **Encapsulation:** Bundling of data and methods operating on that data.  
- **Abstraction:** Hiding internal details; exposing only necessary interfaces.  
- **Inheritance:** Mechanism by which one class (child) shares properties and behaviors of another (parent).  
- **Polymorphism:** Ability to treat different classes through a common interface.  

**Why use OOP?**  
- Improves code organization and reusability.  
- Facilitates modeling of real-world entities.  
- Encourages separation of concerns (each class has a single responsibility).  
- Makes large codebases easier to maintain and extend.


### 1.1 Procedural vs. Object-Oriented Solution (Conceptual)

Below is a simple example: managing information about dogs.  

- **Procedural approach:** We store data in separate variables or dictionaries, and write functions to operate on them.  
- **OOP approach:** We define a `Dog` class, create dog instances, and call methods on those instances.


````python
# Procedural style: using dictionaries and functions
dog1 = {'name': 'Buddy', 'age': 3, 'breed': 'Labrador'}
dog2 = {'name': 'Lucy',  'age': 5, 'breed': 'Beagle'}

def describe_dog(dog):
    print(f"{dog['name']} is a {dog['age']}-year-old {dog['breed']}.")

describe_dog(dog1)
describe_dog(dog2)

print("\n---\n")

# Object-oriented style: define a class
class Dog:
    """A simple Dog class."""
    def __init__(self, name, age, breed):
        # these are instance attributes
        self.name = name
        self.age = age
        self.breed = breed
    
    def describe(self):
        print(f"{self.name} is a {self.age}-year-old {self.breed}.")

# Creating instances
buddy = Dog(name='Buddy', age=3, breed='Labrador')
lucy  = Dog(name='Lucy',  age=5, breed='Beagle')

# Calling methods on those instances
buddy.describe()
lucy.describe()
````

## 2. Python Classes & Objects (12 min)

### 2.1 Defining a Class

- Use the `class` keyword followed by the class name (by convention, PascalCase).  
- Inside, define methods (functions) and attributes (data).  
- The first argument of any instance method is `self`, which refers to the instance.

```python
class MyClass:
    def __init__(self, x):
        self.x = x
    
    def do_something(self):
        return self.x * 2
```

- `MyClass` is a blueprint; to use it, we **instantiate** it:
  ```python
  obj = MyClass(10)
  print(obj.x)           # 10
  print(obj.do_something())  # 20
  ```


````python
# Example: a very basic class with no attributes or methods
class Empty:
    pass

e = Empty()
print(type(e))   # <class '__main__.Empty'>
print(e)         # <__main__.Empty object at ...>

# Adding attributes dynamically (not recommended for production, but possible)
e.name = 'Anonymous'
e.value = 42
print(e.name, e.value)
````

## 3. Attributes vs. Methods (12 min)

### 3.1 Instance Attributes

- Defined inside `__init__` via `self`.  
- Unique to each instance.  
- Accessed via `instance.attribute_name`.

### 3.2 Methods

- Functions defined inside a class.  
- Always take `self` as the first parameter (unless using `@staticmethod` or `@classmethod`).  
- Called via `instance.method_name()`.

### 3.3 Class Attributes

- Defined directly in the class block (outside of any method).  
- Shared among all instances of the class.

```python
class Circle:
    pi = 3.14159   # class attribute

    def __init__(self, radius):
        self.radius = radius  # instance attribute

    def area(self):
        return Circle.pi * (self.radius ** 2)
```

- All `Circle` instances share the same `pi`, but each has its own `radius`.


````python
class Circle:
    pi = 3.14159  # class attribute

    def __init__(self, radius):
        self.radius = radius  # instance attribute

    def area(self):
        return Circle.pi * self.radius ** 2

c1 = Circle(2)
c2 = Circle(5)

print("c1 radius:", c1.radius)
print("c1 area:", c1.area())
print("c2 radius:", c2.radius)
print("c2 area:", c2.area())

# Changing class attribute
Circle.pi = 3.14
print("After changing pi:")
print("c1 area:", c1.area())
print("c2 area:", c2.area())
````

## 4. Constructors & Default Arguments (10 min)

- The `__init__` method can have default arguments so that users don’t always have to provide every parameter.

```python
class Garage:
    def __init__(self, name="My Garage", capacity=10):
        self.name = name
        self.capacity = capacity

    def info(self):
        print(f"Garage '{self.name}' can hold up to {self.capacity} cars.")
```

- **Examples:**
  ```python
  g1 = Garage()             # uses defaults: name="My Garage", capacity=10
  g2 = Garage("SuperCars", 5)
  g1.info()  # Garage 'My Garage' can hold up to 10 cars.
  g2.info()  # Garage 'SuperCars' can hold up to 5 cars.
  ```


````python
# Using default arguments in __init__
class Garage:
    def __init__(self, name="Generic Garage", capacity=5):
        self.name = name
        self.capacity = capacity
        self.cars = []      # list to store parked car names

    def park(self, car_name):
        if len(self.cars) < self.capacity:
            self.cars.append(car_name)
            print(f"Parked {car_name}. Current cars: {self.cars}")
        else:
            print("Garage is full!")

    def leave(self, car_name):
        if car_name in self.cars:
            self.cars.remove(car_name)
            print(f"{car_name} left. Current cars: {self.cars}")
        else:
            print(f"{car_name} not found in the garage.")

    def info(self):
        print(f"Garage '{self.name}' ({len(self.cars)}/{self.capacity})")

# Create instances and demonstrate default vs. custom initialization
default_garage = Garage()
custom_garage = Garage("Elite Park", 2)

default_garage.info()
default_garage.park("Honda Civic")
default_garage.leave("Toyota Camry")  # not in garage

custom_garage.park("Ferrari")
custom_garage.park("Lamborghini")
custom_garage.park("Porsche")  # Should report "Garage is full!"
````

## 5. Class Variables & Class/Static Methods (10 min)

### 5.1 Class Variables (Review)

- Declared in the class body, outside any method.
- Shared among all instances.
- Useful for counting instances, default settings, etc.

```python
class Product:
    total_products = 0  # class variable

    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.total_products += 1
```

- Every time a new `Product` is created, `total_products` increments.

### 5.2 Class Methods (`@classmethod`)

- Receives the class itself as the first argument (`cls`), instead of an instance (`self`).
- Use for factory methods or operations that affect the class as a whole.

```python
class Product:
    total_products = 0

    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.total_products += 1

    @classmethod
    def get_total_products(cls):
        return cls.total_products
```

- Call via `Product.get_total_products()` or `instance.get_total_products()`.

### 5.3 Static Methods (`@staticmethod`)

- Does not receive `self` or `cls`.  
- Behaves like a regular function but lives in the class namespace.

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

MathUtils.add(3, 5)  # 8
```


````python
class Product:
    total_products = 0

    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.total_products += 1

    @classmethod
    def get_total_products(cls):
        return cls.total_products

    @staticmethod
    def is_expensive(price):
        return price > 100

# Create some products
p1 = Product("Keyboard", 50)
p2 = Product("Monitor", 150)
p3 = Product("Laptop", 1200)

print("Total products created:", Product.get_total_products())
print("Is 75 expensive?", Product.is_expensive(75))   # False
print("Is 200 expensive?", Product.is_expensive(200)) # True
````

## 6. Inheritance & Polymorphism (18 min)

### 6.1 Single Inheritance

- A derived (child) class inherits attributes and methods from a base (parent) class.  
- Use syntax: `class Child(Parent):`

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclasses must implement this method.")

class Dog(Animal):
    def speak(self):
        return f"{self.name} says: Woof!"

class Cat(Animal):
    def speak(self):
        return f"{self.name} says: Meow!"
```

- `Dog` and `Cat` inherit `__init__` from `Animal` but override `speak()`.

### 6.2 Overriding & `super()`

- A child class can override a parent’s method.  
- Use `super().__init__(...)` to call the parent constructor.

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)
        self.num_doors = num_doors

    def info(self):
        return f"{self.make} {self.model} with {self.num_doors} doors."
```

### 6.3 Polymorphism (Duck Typing)

- Different classes implementing the same method name can be used interchangeably.

```python
def make_it_speak(animal):
    print(animal.speak())

d = Dog("Buddy")
c = Cat("Whiskers")

make_it_speak(d)  # Buddy says: Woof!
make_it_speak(c)  # Whiskers says: Meow!
```

````python
# Demonstration of inheritance and polymorphism
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclasses must implement this method.")

class Dog(Animal):
    def speak(self):
        return f"{self.name} says: Woof!"

class Cat(Animal):
    def speak(self):
        return f"{self.name} says: Meow!"

class Bird(Animal):
    def speak(self):
        return f"{self.name} says: Tweet!"

# Using polymorphism
animals = [Dog("Buddy"), Cat("Luna"), Bird("Tweety")]
for a in animals:
    print(a.speak())
````

## 7. Encapsulation & Special Methods (15 min)

### 7.1 Encapsulation (Protected & Private Attributes)

- By convention, prefix an attribute with a single underscore `_` to indicate it’s “protected” (internal use).  
- Prefix with double underscore `__` to trigger name mangling, making it harder (but not impossible) to access from outside.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner              # public
        self._transactions = []         # protected
        self.__balance = balance        # private (name-mangled to _BankAccount__balance)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self._transactions.append(("deposit", amount))

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            self._transactions.append(("withdraw", amount))

    def get_balance(self):
        return self.__balance
```

- **Accessing “private” attribute from outside (not recommended):**  
  ```python
  acct = BankAccount("Alice", 100)
  print(acct._BankAccount__balance)  # 100 (but avoid using this)
  ```

### 7.2 Special Methods (Dunder Methods)

- Python uses “magic” or “dunder” methods like `__str__`, `__repr__`, `__add__`, etc.  
- Let you define behavior for built-in operations:

| Method         | Called by                   | Purpose                              |
|----------------|-----------------------------|--------------------------------------|
| `__init__`     | constructor (object creation) | Initialize instance attributes       |
| `__str__`      | `str(obj)` or `print(obj)`    | Human-readable string representation |
| `__repr__`     | `repr(obj)`                   | Unambiguous string (for debugging)   |
| `__add__`      | `obj1 + obj2`                 | Define addition behavior             |
| `__eq__`       | `obj1 == obj2`                | Define equality check                |
| `__lt__`, etc. | `<, >, <=, >=`                | Define ordering                      |


````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner              # public
        self._transactions = []         # protected (convention)
        self.__balance = balance        # private (name-mangled)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self._transactions.append(("deposit", amount))

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            self._transactions.append(("withdraw", amount))

    def get_balance(self):
        return self.__balance

acct = BankAccount("Alice", 100)
acct.deposit(50)
acct.withdraw(30)
print("Balance:", acct.get_balance())
print("Transactions:", acct._transactions)

try:
    print(acct.__balance)
except AttributeError as e:
    print("AttributeError:", e)

print("Name-mangled access (avoid!):", acct._BankAccount__balance)
````

````python
class Money:
    def __init__(self, dollars, cents):
        self.dollars = dollars
        self.cents = cents

    def __str__(self):
        return f"${self.dollars}.{self.cents:02d}"

    def __repr__(self):
        return f"Money({self.dollars}, {self.cents})"

    def __add__(self, other):
        total_cents = (self.dollars * 100 + self.cents) + (other.dollars * 100 + other.cents)
        return Money(total_cents // 100, total_cents % 100)

    def __eq__(self, other):
        return (self.dollars, self.cents) == (other.dollars, other.cents)

# Test special methods
m1 = Money(10, 75)
m2 = Money(5, 50)
print("m1:", m1)           # __str__
print("repr(m1):", repr(m1))  # __repr__
sum_money = m1 + m2         # __add__
print("m1 + m2:", sum_money)
print("m1 == m2?", m1 == m2)  # __eq__
print("m1 == Money(10,75)?", m1 == Money(10, 75))
````

## 8. Composition & Advanced Features (10 min)

### 8.1 Composition (Has-A Relationship)

- Instead of inheriting, a class can **contain** instances of other classes.
- Useful when “is-a” doesn’t make sense but “has-a” does.

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

class Car:
    def __init__(self, make, model, horsepower):
        self.make = make
        self.model = model
        self.engine = Engine(horsepower)

    def specs(self):
        return f"{self.make} {self.model} with {self.engine.horsepower} HP"
```

### 8.2 The `@property` Decorator

- Provides a way to define getters/setters without explicit methods.
- Makes attribute access syntax cleaner.

```python
class Rectangle:
    def __init__(self, width, height):
        self._width = width
        self._height = height

    @property
    def area(self):
        return self._width * self._height

    @area.setter
    def area(self, value):
        raise AttributeError("Cannot set area directly; modify width or height instead.")
```

- Access via `rect.area` (no parentheses), but it’s computed on the fly.

### 8.3 Other Advanced Topics (Brief Overview)

- **Decorators on methods** (e.g., `@staticmethod`, `@classmethod`).  
- **Meta-classes**: specify behavior of class creation.  
- **Data classes (`@dataclass`)**: automatic `__init__`, `__repr__`, etc.  
- **SOLID principles** and design patterns (Factory, Singleton, Observer).


````python
# Composition example
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

class Car:
    def __init__(self, make, model, horsepower):
        self.make = make
        self.model = model
        self.engine = Engine(horsepower)

    def specs(self):
        return f"{self.make} {self.model} with {self.engine.horsepower} HP"

c = Car("Tesla", "Model S", 1020)
print(c.specs())

# @property example
class Rectangle:
    def __init__(self, width, height):
        self._width = width
        self._height = height

    @property
    def area(self):
        return self._width * self._height

    @area.setter
    def area(self, value):
        raise AttributeError("Cannot set area directly; modify width or height instead.")

r = Rectangle(3, 5)
print("Area of rectangle:", r.area)
````

## 10. Next Steps & Best Practices (5 min)

- **Design Principles:**  
  - **Single Responsibility Principle**: Each class should have one reason to change.  
  - **Open/Closed Principle**: Classes should be open for extension but closed for modification.  
  - **Liskov Substitution Principle**: Subclasses should be usable wherever their base class is expected.  
  - **Interface Segregation Principle**: Clients should not be forced to depend on methods they do not use.  
  - **Dependency Inversion Principle**: High-level modules should not depend on low-level modules; use abstractions.

- **Documentation & Style:**  
  - Always write docstrings for classes and methods.  
  - Follow PEP 8 naming conventions: `PascalCase` for classes, `snake_case` for methods/functions and variables.  
  - Use meaningful names and keep classes focused on a single responsibility.

- **Real-World Practice:**  
  - Explore design patterns (Factory, Singleton, Observer, Strategy, etc.).  
  - Learn about `@dataclass` for boilerplate reduction.  
  - Dive into metaclasses if you need to customize class creation.  
  - Experiment with frameworks that heavily utilize OOP (Django, Flask, etc.).  
  - Build small projects: e.g., a simple inventory system, a game with objects, etc.

````python
# Solution for Exercise 3

class BankAccount:
    """A simple bank account class."""
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            return f"Deposited ${amount}. New balance: ${self.balance}"
        return "Invalid deposit amount"

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            return f"Withdrew ${amount}. New balance: ${self.balance}"
        return "Insufficient funds or invalid amount"

    def get_balance(self):
        return self.balance

# Create instance and run scenario
acct = BankAccount("Alice", 100)
print(acct.deposit(50))      # Deposited $50. New balance: $150
print(acct.withdraw(30))     # Withdrew $30. New balance: $120
print("Final balance:", acct.get_balance())  # 120
````

````python
# Solution for Exercise 6

class Money:
    """A simple Money class with operator overloading."""
    def __init__(self, dollars, cents):
        self.dollars = dollars
        self.cents = cents

    def __str__(self):
        # Ensure cents are two digits
        return f"${self.dollars}.{self.cents:02d}"

    def __repr__(self):
        return f"Money({self.dollars}, {self.cents})"

    def __add__(self, other):
        total_cents = (self.dollars * 100 + self.cents) + (other.dollars * 100 + other.cents)
        return Money(total_cents // 100, total_cents % 100)

    def __eq__(self, other):
        if not isinstance(other, Money):
            return False
        return (self.dollars, self.cents) == (other.dollars, other.cents)

# Demonstration
m1 = Money(5, 75)
m2 = Money(3, 50)
print("m1:", m1)
print("m2:", m2)
print("m1 + m2 =", m1 + m2)            # Expect $9.25
print("m1 == m2?", m1 == m2)           # False
print("m1 == Money(5,75)?", m1 == Money(5, 75))  # True
````

### Exercise 7 (Stretch, 25 min): Library Management System

**Prompt:**  
Design a mini library system with the following classes and relationships:

1. **`Book`**  
   - Attributes: `title` (str), `author` (str), `isbn` (str), `available_copies` (int).  
   - Methods:  
     - `__str__()` → returns a readable string.  
     - `is_available()` → returns `True` if `available_copies > 0`, else `False`.  
     - `borrow()` → decrements `available_copies` by 1 if available, else returns `"No copies available."`.  
     - `return_book()` → increments `available_copies` by 1.

2. **`Member`**  
   - Attributes: `name` (str), `member_id` (str), `borrowed_books` (list of `Book` instances).  
   - Methods:  
     - `borrow_book(book: Book)` → if `book.is_available()`, call `book.borrow()` and add to `borrowed_books`. Else, print a message.  
     - `return_book(book: Book)` → if `book` in `borrowed_books`, call `book.return_book()` and remove from `borrowed_books`.

3. **`Library`**  
   - Attributes: `books` (list of `Book`), `members` (list of `Member`).  
   - Methods:  
     - `add_book(book: Book)` → adds to `books`.  
     - `add_member(member: Member)` → adds to `members`.  
     - `find_book_by_title(title: str)` → returns the first matching `Book` or `None`.  
     - `lend_book(title: str, member_id: str)` → finds the book and member, then calls the appropriate methods.  
     - `receive_return(title: str, member_id: str)` → reverse of lending.

**After defining all classes, demonstrate**:  
- Creating a few books and a few members.  
- Adding them to the library.  
- Member borrows a book, then returns it.


````python
# Solution for Exercise 7

class Book:
    """Represents a book in the library."""
    def __init__(self, title, author, isbn, available_copies):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.available_copies = available_copies

    def __str__(self):
        return f"\"{self.title}\" by {self.author} (ISBN: {self.isbn}) - Copies: {self.available_copies}"

    def is_available(self):
        return self.available_copies > 0

    def borrow(self):
        if self.is_available():
            self.available_copies -= 1
            return True
        return False

    def return_book(self):
        self.available_copies += 1

class Member:
    """Represents a library member."""
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []

    def borrow_book(self, book: Book):
        if book.is_available():
            book.borrow()
            self.borrowed_books.append(book)
            print(f"{self.name} borrowed \"{book.title}\".")
        else:
            print(f"Sorry, \"{book.title}\" is not available.")

    def return_book(self, book: Book):
        if book in self.borrowed_books:
            book.return_book()
            self.borrowed_books.remove(book)
            print(f"{self.name} returned \"{book.title}\".")
        else:
            print(f"{self.name} did not borrow \"{book.title}\".")

class Library:
    """Represents the library itself."""
    def __init__(self):
        self.books = []
        self.members = []

    def add_book(self, book: Book):
        self.books.append(book)

    def add_member(self, member: Member):
        self.members.append(member)

    def find_book_by_title(self, title: str):
        for book in self.books:
            if book.title.lower() == title.lower():
                return book
        return None

    def find_member_by_id(self, member_id: str):
        for member in self.members:
            if member.member_id == member_id:
                return member
        return None

    def lend_book(self, title: str, member_id: str):
        book = self.find_book_by_title(title)
        member = self.find_member_by_id(member_id)
        if not book:
            print(f"Book \"{title}\" not found in library.")
            return
        if not member:
            print(f"Member ID '{member_id}' not found.")
            return
        member.borrow_book(book)

    def receive_return(self, title: str, member_id: str):
        book = self.find_book_by_title(title)
        member = self.find_member_by_id(member_id)
        if not book:
            print(f"Book \"{title}\" not found in library.")
            return
        if not member:
            print(f"Member ID '{member_id}' not found.")
            return
        member.return_book(book)

# Demonstration
lib = Library()

# Create some books
b1 = Book("The Hobbit", "J.R.R. Tolkien", "978-0547928227", 3)
b2 = Book("To Kill a Mockingbird", "Harper Lee", "978-0061120084", 2)
b3 = Book("1984", "George Orwell", "978-0451524935", 1)

# Add books to library
lib.add_book(b1)
lib.add_book(b2)
lib.add_book(b3)

# Create members
m1 = Member("Alice", "M001")
m2 = Member("Bob",   "M002")

# Add members to library
lib.add_member(m1)
lib.add_member(m2)

# Alice borrows "1984"
lib.lend_book("1984", "M001")
# Alice tries to borrow "1984" again (no copies left)
lib.lend_book("1984", "M001")

# Bob tries to borrow "1984" (should be unavailable)
lib.lend_book("1984", "M002")

# Bob borrows "The Hobbit"
lib.lend_book("The Hobbit", "M002")

# Alice returns "1984"
lib.receive_return("1984", "M001")

# Bob borrows "1984" now that it's returned
lib.lend_book("1984", "M002")
````

````python
# Solution for Exercise 8

class Temperature:
    """A temperature stored internally in Celsius."""
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def fahrenheit(self):
        return self.celsius * 9 / 5 + 32

    @property
    def kelvin(self):
        return self.celsius + 273.15

    @classmethod
    def from_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)

    @staticmethod
    def is_freezing(celsius):
        return celsius <= 0

t1 = Temperature(25)
t2 = Temperature.from_fahrenheit(212)

for t in [t1, t2]:
    print(f"{t.celsius:.2f} °C = {t.fahrenheit:.2f} °F = {t.kelvin:.2f} K")

print("Is -5 °C freezing?", Temperature.is_freezing(-5))
print("Is 25 °C freezing?", Temperature.is_freezing(t1.celsius))
````

````python
# Solution for Exercise 9

class Student:
    """A student with a list of grades (0-20 scale)."""
    def __init__(self, name):
        self.name = name
        # A new list for EACH student. A mutable default like grades=[]
        # would be shared by every student created without grades!
        self.grades = []

    def add_grade(self, grade):
        if not 0 <= grade <= 20:
            raise ValueError(f"Grade must be between 0 and 20, got {grade}")
        self.grades.append(grade)

    def average(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)

    def __str__(self):
        return f"{self.name}: average {self.average():.2f} ({len(self.grades)} grades)"

    def __lt__(self, other):
        return self.average() < other.average()

ana = Student("Ana")
for g in [14, 16, 16]:
    ana.add_grade(g)

ben = Student("Ben")
for g in [12, 11]:
    ben.add_grade(g)

carla = Student("Carla")
for g in [19, 18, 20, 17]:
    carla.add_grade(g)

for s in sorted([ana, ben, carla]):
    print(s)

try:
    ana.add_grade(25)
except ValueError as e:
    print("Error:", e)
````

````python
# Solution for Exercise 11

class Thermostat:
    MIN_TEMP = 10
    MAX_TEMP = 30

    def __init__(self, target):
        self.target = target          # goes through the setter -> validated

    @property
    def target(self):
        return self._target

    @target.setter
    def target(self, value):
        if not self.MIN_TEMP <= value <= self.MAX_TEMP:
            raise ValueError(f"Target must be between {self.MIN_TEMP} and {self.MAX_TEMP}, got {value}")
        self._target = value

t = Thermostat(20)
print("Initial target:", t.target)
t.target = 22
print("New target:", t.target)

for bad in [35]:
    try:
        t.target = bad
    except ValueError as e:
        print("Error:", e)

try:
    Thermostat(5)
except ValueError as e:
    print("Error when creating:", e)

print("Target unchanged after the error:", t.target)
````

### Exercise 12 (20 min): Shopping Cart (Composition)

**Prompt:**

- Create a class `ShoppingCart` that **has** a list of `Item` objects (composition: a cart *has* items, it *is not* an item).
  - `add(item)`: adds an item. If `item` is not an `Item`, raise a `TypeError` (hint: `isinstance`).
  - `remove(name)`: removes the item with that name. If it is not in the cart, raise a `KeyError`.
  - `total(discount=0.0)`: returns the sum of the subtotals with an optional discount between 0 and 1 (e.g. `0.1` = 10% off).
  - `__len__()`: returns the **total number of units**, so `len(cart)` works.
  - `__str__()`: shows one line per item and the total.
- Test it: add 3 items, print the cart, `len(cart)`, the total with a 10% discount, then remove one item and try to add the string `"banana"`.

```python
# TODO: Write classes Item and ShoppingCart here
```

````python
# Solution for Exercise 12

class Item:
    def __init__(self, name, price, quantity=1):
        self.name = name
        self.price = price
        self.quantity = quantity

    def subtotal(self):
        return self.price * self.quantity

    def __str__(self):
        return f"{self.quantity} x {self.name} @ {self.price:.2f} € = {self.subtotal():.2f} €"

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add(self, item):
        if not isinstance(item, Item):
            raise TypeError(f"Can only add Item objects, got {type(item).__name__}")
        self.items.append(item)

    def remove(self, name):
        for item in self.items:
            if item.name == name:
                self.items.remove(item)
                return
        raise KeyError(f"No item named {name!r} in the cart")

    def total(self, discount=0.0):
        if not 0 <= discount <= 1:
            raise ValueError("discount must be between 0 and 1")
        return sum(item.subtotal() for item in self.items) * (1 - discount)

    def __len__(self):
        return sum(item.quantity for item in self.items)

    def __str__(self):
        lines = [str(item) for item in self.items]
        lines.append(f"TOTAL: {self.total():.2f} €")
        return "\n".join(lines)

cart = ShoppingCart()
cart.add(Item("Coffee", 4.50, 2))
cart.add(Item("Croissant", 1.80, 3))
cart.add(Item("Water", 1.20))

print(cart)
print("Units in cart:", len(cart))
print(f"Total with 10% discount: {cart.total(0.1):.2f} €")

cart.remove("Water")
print("After removing Water:", len(cart), "units")

try:
    cart.add("banana")
except TypeError as e:
    print("TypeError:", e)

try:
    cart.remove("Pizza")
except KeyError as e:
    print("KeyError:", e)
````

  - `__abs__` → the length (norm) √(x² + y²), so `abs(v)` works
  - `__eq__` → two vectors are equal if their `x` and `y` are equal
  - a method `dot(other)` → the dot product x₁·x₂ + y₁·y₂

**Why it matters:** this is how NumPy arrays let you write `a + b` and `a * 2`. They are just classes with dunder methods!

```python
# TODO: Write class Vector2D here
```

````python
# Solution for Exercise 13
import math

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector2D(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector2D(self.x * scalar, self.y * scalar)

    def __abs__(self):
        return math.sqrt(self.x ** 2 + self.y ** 2)

    def __eq__(self, other):
        if not isinstance(other, Vector2D):
            return False
        return self.x == other.x and self.y == other.y

    def dot(self, other):
        return self.x * other.x + self.y * other.y

v = Vector2D(3, 4)
w = Vector2D(1, 2)
print("v + w =", v + w)
print("v - w =", v - w)
print("v * 2 =", v * 2)
print("|v|   =", abs(v))
print("v . w =", v.dot(w))
print("v == Vector2D(3, 4)?", v == Vector2D(3, 4))
````

````python
# Solution for Exercise 15
import math

class MeanScaler:
    """Standardise numbers: z = (x - mean) / std. Mimics scikit-learn's StandardScaler."""
    def __init__(self):
        self.mean_ = None
        self.std_ = None

    def fit(self, data):
        n = len(data)
        self.mean_ = sum(data) / n
        self.std_ = math.sqrt(sum((x - self.mean_) ** 2 for x in data) / n)
        return self                       # returning self allows chaining

    def _check_fitted(self):
        if self.mean_ is None:
            raise RuntimeError("MeanScaler is not fitted yet. Call fit() first.")

    def transform(self, data):
        self._check_fitted()
        if self.std_ == 0:
            return [0.0 for _ in data]
        return [(x - self.mean_) / self.std_ for x in data]

    def fit_transform(self, data):
        return self.fit(data).transform(data)

    def inverse_transform(self, z_values):
        self._check_fitted()
        return [z * self.std_ + self.mean_ for z in z_values]

    def __repr__(self):
        return f"MeanScaler(mean_={self.mean_}, std_={self.std_})"

train = [10, 20, 30, 40, 50]
test = [25, 60]

scaler = MeanScaler()
z_train = scaler.fit_transform(train)
print(scaler)
print("z_train:", [round(z, 3) for z in z_train])

m = sum(z_train) / len(z_train)
s = math.sqrt(sum((z - m) ** 2 for z in z_train) / len(z_train))
print(f"mean of z_train = {m:.3f}, std of z_train = {s:.3f}")

z_test = scaler.transform(test)       # uses TRAIN mean/std: no peeking at test data
print("z_test:", [round(z, 3) for z in z_test])
print("back to original:", [round(x, 6) for x in scaler.inverse_transform(z_test)])

try:
    MeanScaler().transform([1, 2, 3])
except RuntimeError as e:
    print("RuntimeError:", e)

# Why not re-fit on test data? The model must treat new data exactly like the
# training data. Re-fitting would use information from the test set
# ("data leakage") and the same raw value would map to a different z-score.
````
