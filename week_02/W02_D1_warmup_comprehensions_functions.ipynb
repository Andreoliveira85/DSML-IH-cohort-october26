## 1 · List comprehensions

### From a loop to one line
Very often a loop just **builds a new list** from an old one. Here we convert steps into thousands of steps:

````python
# The loop way: 3 lines
steps_k = []
for s in steps:
    steps_k.append(s / 1000)

print(steps_k)
````

A **list comprehension** does exactly the same in **one line**:

```python
new_list = [expression for item in collection]
```

Read it from the inside out, like English: *"give me `s / 1000` **for each** `s` **in** `steps`"*.

````python
steps_k = [s / 1000 for s in steps]
print(steps_k)
````

### Filter: keep only some elements
Add an `if` **at the end** to keep only the items that pass a test:

```python
new_list = [expression for item in collection if condition]
```

Which days did the person walk **at least 8,000 steps**?

````python
active_steps = [s for s in steps if s >= 8000]
print(active_steps)
print("Active days:", len(active_steps))
````

### Choose between two values: `if ... else` **before** the `for`
To **label** every element (instead of throwing some away), put `value_if_true if condition else value_if_false` at the **start**:

````python
labels = ["active" if s >= 8000 else "rest" for s in steps]
print(labels)
````

> 💡 **Where does the `if` go?**
> - `[x for x in data if x > 0]` → **filter**: the new list can be **shorter**.
> - `[a if x > 0 else b for x in data]` → **choose**: the new list has the **same length**.

## 2 · Function example (easy): steps → kilometres

A smartwatch estimates distance as **steps × stride length**. An average stride is about **0.75 m**.

Let's package this in a function with a **default value** for the stride:

````python
def steps_to_km(n_steps, stride_m=0.75):
    """Return the distance in km walked with n_steps steps of stride_m metres."""
    distance_m = n_steps * stride_m
    return round(distance_m / 1000, 2)

print(steps_to_km(10000))                 # default stride 0.75 m
print(steps_to_km(10000, stride_m=0.80))  # a taller person
````

Anatomy, one more time:
- `def` + **name** + **arguments** in brackets, then `:`
- `stride_m=0.75` is a **default value**: used when the caller doesn't give one
- the **docstring** `"""..."""` says what the function does
- `return` **gives the result back** so we can store it and keep calculating

### ✨ Functions + comprehensions = a great team
Now we can compute the distance for **every day** in one line:

````python
km_per_day = [steps_to_km(s) for s in steps]
print(km_per_day)
print("Total this week:", round(sum(km_per_day), 2), "km")
````

## 3 · 🧑‍💻 Your turn: weekly activity summary (5–7 min)

Write a function `weekly_summary(steps, goal=8000)` that **returns** three values:
1. the **average** number of steps per day, rounded to a whole number
2. the **number of days** on which the goal was reached (`>= goal`)
3. the **percentage** of days on which the goal was reached, rounded to 1 decimal

Then test it with our `steps` list, once with the default goal and once with `goal=10000`.

**Expected output:**
```
(8357, 4, 57.1)
(8357, 2, 28.6)
```

💡 **Hints**
- `sum(...)` and `len(...)` give you the average.
- Count the good days with a **list comprehension** + `len(...)`.
- A function can return several values: `return a, b, c` (Python packs them into a **tuple**).

````python
def weekly_summary(steps, goal=8000):
    """Return (average steps, days goal reached, % of days goal reached)."""
    average = round(sum(steps) / len(steps))
    good_days = len([s for s in steps if s >= goal])
    percent = round(good_days / len(steps) * 100, 1)
    return average, good_days, percent

print(weekly_summary(steps))
print(weekly_summary(steps, goal=10000))

# Unpack the three returned values into three variables
avg, n_days, pct = weekly_summary(steps)
print(f"Average {avg} steps/day · goal reached on {n_days} days ({pct}%)")
````

### ⭐ Bonus (if you finish early): *which* days?
Write `goal_days(days, steps, goal=8000)` that returns the **names** of the days on which the goal was reached → `['Tue', 'Wed', 'Fri', 'Sat']`.

💡 **New tool: `zip`** walks through two lists **side by side**, giving you one pair at a time:
```python
for d, s in zip(days, steps):
    print(d, s)     # Mon 6200, Tue 8450, ...
```

````python
def goal_days(days, steps, goal=8000):
    """Return the names of the days with at least `goal` steps."""
    return [d for d, s in zip(days, steps) if s >= goal]

print(goal_days(days, steps))
print(goal_days(days, steps, goal=10000))
````

## 🧭 Recap

| Pattern | Example |
|---|---|
| transform | `[s / 1000 for s in steps]` |
| filter | `[s for s in steps if s >= 8000]` |
| choose | `["active" if s >= 8000 else "rest" for s in steps]` |
| function + default | `def steps_to_km(n_steps, stride_m=0.75):` |
| function inside a comprehension | `[steps_to_km(s) for s in steps]` |
| return several values | `return average, good_days, percent` |
